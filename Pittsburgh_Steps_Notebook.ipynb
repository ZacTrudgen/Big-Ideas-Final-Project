The program below finds how often a neighborhood is refferenced in the data *(aka how many staircases are in that neighborhood)*
- Data is collected from the "neighborhood" column and then stored in a dictionary
- The data is then turned into a graph for presentation

# Most Climbable Neighborhood in Pittsburgh (Public Staircases data):

The code below counts how many times each neighborhood is refferenced in the data. This directly reflects how many staircases are in the given neighborhood.  

## Neighborhood with the most Staircases:

The code below counts how many times each neighborhood is refferenced in the data. This directly reflects how many staircases are in the given neighborhood.  

```python
import pandas as pd

data = pd.read_csv("Pittsburgh_Steps_Data.csv", sep=",")

data.value_counts("neighborhood").sort_index().sort_values().plot.bar(ylabel = "Staircases", figsize = (15, 5), width = (0.70), title = "Staircases per Neighborhood", color='blue')
```

## Neighborhood with the most Individual Steps:

The code below groups the data from the csv file by neighborhood; then it adds together the total number of steps in each neighborhood group.

```python
import pandas as pd

data = pd.read_csv("Pittsburgh_Steps_Data.csv", sep=",")

data.groupby(by="neighborhood")["number_of_steps"].sum().sort_values().plot.bar(ylabel = "Steps", figsize = (15, 5), width = (0.70), title = "Steps per Neighborhood", color='red')
```

## Neighborhood with the most Vertical Hight:

This code multiplies each step by 7 (average public staircase height), to estimate the total hight you can climb in each neighborhood

```python
import pandas as pd

data = pd.read_csv("Pittsburgh_Steps_Data.csv", sep=",")

data["number_of_steps"] = data["number_of_steps"] * 7 

data.groupby(by="neighborhood")["number_of_steps"].sum().sort_values().plot.bar(ylabel = "Height (ft)", figsize = (15, 5), width = (0.70), title = "Height(ft) per Neighborhood", color='black')
```
